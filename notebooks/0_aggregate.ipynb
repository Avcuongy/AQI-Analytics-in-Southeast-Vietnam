```python
import pandas as pd
```

```python
dfs = [
    pd.read_csv(f"../data/historical/air_quality/air_quality_year_{year}.csv")
    for year in range(2020, 2027)
]
df_aqi = pd.concat(dfs, axis=0, ignore_index=True)
df_aqi.head()
```

```python
df_aqi.info()
```

---

```python
dfs = [
    pd.read_csv(f"../data/historical/weather/weather_year_{year}.csv")
    for year in range(2020, 2027)
]
df_weather = pd.concat(dfs, axis=0, ignore_index=True)
df_weather.head()
```

```python
df_weather.info()
```

```python
df = pd.merge(
    df_aqi, df_weather, on=["date", "location_id", "location_name"], how="inner"
)
```

```python
df.info()
```

```python
df.head()
```

```python
df.dropna(inplace=True)
df.info()
```

```python
df.to_csv(
    "../data/clean/data.csv",
    index=False,
)
```
